```python
feature_columns = [
    "danceability",
    "energy",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence"
]

for column in feature_columns:
    invalid_count = ((df[column] < 0) | (df[column] > 1)).sum()
    print(f"{column}: {invalid_count} invalid values")
```
